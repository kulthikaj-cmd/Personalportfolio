ACTIVITY 1.2

```python
width = float(input("Enter width (m): "))
length = float(input("Enter length (m): "))

area = width * length

print("Width =", width)
print("Length =", length)
print("Area =", area, "square meters")

print("Data Type of area:", type(area))
```

Data Type

width = float
length = float
area = float

เหตุผล

โปรแกรมรับค่าความกว้างและความยาวจากผู้ใช้โดยใช้ฟังก์ชัน float() เพื่อให้สามารถรับค่าที่เป็นเลขทศนิยมได้ จากนั้นนำค่าทั้งสองมาคูณกันเพื่อหาพื้นที่ของสี่เหลี่ยมผืนผ้า ผลลัพธ์จะถูกเก็บไว้ในตัวแปร area ซึ่งมีชนิดข้อมูลเป็น float เนื่องจากค่าที่นำมาคำนวณเป็นข้อมูลชนิด float ทำให้ผลลัพธ์ที่ได้เป็น float เช่นกัน
