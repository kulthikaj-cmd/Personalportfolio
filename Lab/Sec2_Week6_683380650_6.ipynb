Lab 6.1: Simple Calculator Functions

````python

def is_positive_negative_zero(num):
    """Identify whether a number is positive, negative, or zero."""
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"


def is_even_odd(num):
    """Identify whether an integer is even or odd."""
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"


number = int(input("Enter an integer: "))

print("Number type:", is_positive_negative_zero(number))
print("Even or Odd:", is_even_odd(number))
````

Lab 6.2: Custom Module and Standard Library Usage

````python

%%writefile my_utils.py

def greet(name):
    """Print a greeting message."""
    print(f"Hello, {name}!")


def is_prime(number):
    """Return True if the number is prime, otherwise False."""
    if number < 2:
        return False

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True
````

````python

import my_utils
import math
import random

my_utils.greet("Alice")

print("Is 7 prime?", my_utils.is_prime(7))
print("Is 10 prime?", my_utils.is_prime(10))

number = 25
print("Square root of", number, "is", math.sqrt(number))

random_number = random.randint(1, 100)
print("Random number:", random_number)
````

## Activity 6.2: Benefits of Modular Programming & Standard Library Exploration

### 1. Benefits of Modular Programming

Modular programming divides a large program into smaller functions and modules. It makes code easier to understand, reuse, test, debug, and maintain. It also helps team members work on different parts of a project.

For example, in a student management system, separate functions can be used to add, search, update, and delete student information.

### 2. Python Standard Library Exploration

Official documentation: https://docs.python.org/3/library/

**Module 1: datetime**

The `datetime` module is used to work with dates and times. For example, it can display the current date.

```python
import datetime

print(datetime.datetime.now())
```

**Module 2: random**

The `random` module is used to generate random values. For example, it can generate a random number from 1 to 10.

```python
import random

print(random.randint(1, 10))
```

**Module 3: math**

The `math` module provides mathematical functions. For example, it can calculate a square root.

```python
import math

print(math.sqrt(49))
```

These modules provide ready-made functions that help programmers write code more efficiently without implementing everything from scratch.
