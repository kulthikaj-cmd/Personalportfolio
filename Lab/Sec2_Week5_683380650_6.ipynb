Lab 5.1: Simple Contact Book Manager

```python

print("--- Lab 5.1: Contact Book Manager ---")

contacts = {}

while True:
    print("\n[1] Add/Update Contact")
    print("[2] View Contact Details")
    print("[3] List All Contacts")
    print("[4] Delete Contact")
    print("[5] Exit")

    choice = input("Choose a menu (1-5): ")

    if choice == "1":
        name = input("Enter contact name: ").strip().title()
        phone = input("Enter phone number: ")
        email = input("Enter email: ")

        contacts[name] = {
            "phone": phone,
            "email": email
        }

        print(f"Contact '{name}' saved.")

    elif choice == "2":
        name = input("Enter contact name: ").strip().title()
        details = contacts.get(name)

        if details is not None:
            print(f"Name: {name}")
            print(f"Phone: {details['phone']}")
            print(f"Email: {details['email']}")
        else:
            print("Contact not found.")

    elif choice == "3":
        if not contacts:
            print("No contacts found.")
        else:
            for name, details in contacts.items():
                print(f"{name}: {details}")

    elif choice == "4":
        name = input("Enter contact name to delete: ").strip().title()
        deleted = contacts.pop(name, None)

        if deleted is not None:
            print(f"Contact '{name}' deleted.")
        else:
            print("Contact not found.")

    elif choice == "5":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please choose 1-5.")
```

Lab 5.2: Course Enrollment & Collaboration

```python

print("--- Lab 5.2: Course Enrollment ---")

python_students = {"S001", "S003", "S005", "S007", "S009"}
java_students = {"S002", "S003", "S006", "S007", "S010"}

both_courses = python_students.intersection(java_students)
print("Students in both courses:", both_courses)

all_students = python_students.union(java_students)
print("All unique students:", all_students)

python_only = python_students.difference(java_students)
print("Students only in Python:", python_only)

one_course_only = python_students.symmetric_difference(java_students)
print("Students in only one course:", one_course_only)

all_students_list = ["S001", "S003", "S005", "S001", "S002"]
unique_students = list(set(all_students_list))

print("Original list:", all_students_list)
print("Unique list:", unique_students)
```

**1. Where would you use a dictionary in a game?**

A dictionary can store a player's inventory and character statistics. The key identifies each item, and the value stores its information. It is easy to access and update the data.

**2. How could a dictionary represent data about a product in an online store?**

A dictionary can store product information using the product ID as the key. The value can contain the product name, price, and stock quantity. This makes product information easy to find and update.

**3. When would a set be useful in managing user permissions or tags?**

A set can store unique user permissions or hashtags without duplicates. It is useful for checking whether a particular permission or tag exists.

Activity 5.2: Word Frequency Counter

```python

print("--- Activity 5.2: Word Frequency Counter ---")

sentence = input("Enter a sentence: ")
sentence = sentence.lower()

words = sentence.split()
word_counts = {}

for word in words:
    word = word.strip('.,!?"')

    if word == "":
        continue

    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print("Word counts:", word_counts)
```
