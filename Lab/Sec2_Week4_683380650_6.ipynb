lab4_1.py - Student Management System (List Operations)

```python

print("--- Lab 4.1: Student Management System ---")

student_names = []

print("\n[1] Add Students")
for i in range(3):
    name = input(f"Enter student name #{i + 1}: ")
    student_names.append(name)

print(f"Current list: {student_names}")

print("\n[2] Find a Student")
search_name = input("Enter a name to search for: ")

if search_name in student_names:
    position = student_names.index(search_name)
    print(f"Found '{search_name}' at index {position}")
else:
    print(f"'{search_name}' not found in the list")

print("\n[3] Remove a Student")
remove_name = input("Enter a name to remove: ")

if remove_name in student_names:
    student_names.remove(remove_name)
    print(f"'{remove_name}' has been removed")
else:
    print(f"'{remove_name}' not found, nothing removed")

print(f"Updated list: {student_names}")

print("\n[4] Sort Students")
student_names.sort()
print(f"Sorted list: {student_names}")

print("\n[5] Count Students")
print(f"Total students: {len(student_names)}")
```

lab4_2.py - Geometric Calculations (Tuples & Immutability)

```python
import math

print("--- Lab 4.2: Geometric Calculations ---")

point1 = (3, 4)
point2 = (6, 8)

print("\n[1] Access Coordinates")
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

print("\n[2] Attempt Modification (Immutability Demonstration)")
try:
    point1[0] = 5
except TypeError as e:
    print(f"Error trying to modify tuple: {e}")

print("\n[3] Calculate Distance")
distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)
print(f"Distance between point1 {point1} and point2 {point2}: {distance}")
```

Activity 4.1: List vs. Tuple

**1. Daily temperatures recorded over one week — List**

A List is suitable because temperature data may need to be updated or changed.

**2. RGB color values of a specific color — Tuple**

A Tuple is suitable because the RGB values represent a fixed color and should not be changed.

**3. Shopping cart products — List**

A List is suitable because users can add or remove products from the shopping cart.

**4. Days of the week in order — Tuple**

A Tuple is suitable because the days of the week form a fixed sequence.

**5. Student assignment scores — List**

A List is suitable because scores may need to be updated or corrected later.

**6. GPS coordinates of a fixed landmark — Tuple**

A Tuple is suitable because the coordinates represent a fixed location and are not expected to change.

Activity 4.2: Interactive To-Do List

```python

print("--- Activity 4.2: Interactive To-Do List ---")

todo_list = []

while True:
    print("\n[1] Add a task")
    print("[2] View tasks")
    print("[3] Complete a task")
    print("[4] Exit")

    choice = input("Choose a menu (1-4): ")

    if choice == "1":
        task = input("Enter a task: ")
        todo_list.append(task)
        print(f"Added: {task}")

    elif choice == "2":
        if len(todo_list) == 0:
            print("No tasks yet.")
        else:
            print("\nYour To-Do List:")
            for i, task in enumerate(todo_list):
                print(f"{i + 1}. {task}")
            print(f"Total tasks: {len(todo_list)}")

    elif choice == "3":
        if len(todo_list) == 0:
            print("No tasks to complete.")
        else:
            for i, task in enumerate(todo_list):
                print(f"{i + 1}. {task}")

            try:
                task_number = int(input("Enter task number to complete: "))
                if 1 <= task_number <= len(todo_list):
                    completed_task = todo_list.pop(task_number - 1)
                    print(f"Completed: {completed_task}")
                else:
                    print("Invalid task number.")
            except ValueError:
                print("Please enter a number.")

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please choose 1-4.")
```

Challenge ของ Lab 4.2: รับพิกัดจากผู้ใช้

```python


import math

print("--- Lab 4.2: Distance Calculator ---")

x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

point1 = (x1, y1)
point2 = (x2, y2)

distance = math.sqrt(
    (x2 - x1) ** 2 + (y2 - y1) ** 2
)

print("Point 1:", point1)
print("Point 2:", point2)
print("Distance:", distance)
```

Challenge ของ Lab 4.1: Student Management System แบบมีเมนู

```python

print("--- Student Management System ---")

student_names = []

while True:
    print("\n[1] Add Student")
    print("[2] Remove Student")
    print("[3] Find Student")
    print("[4] Sort Students")
    print("[5] Exit")

    choice = input("Choose a menu (1-5): ")

    if choice == "1":
        name = input("Enter student name: ").strip()
        student_names.append(name)
        print("Student added.")

    elif choice == "2":
        name = input("Enter a name to remove: ").strip()
        if name in student_names:
            student_names.remove(name)
            print("Student removed.")
        else:
            print("Student not found.")

    elif choice == "3":
        name = input("Enter a name to search: ").strip()
        if name in student_names:
            position = student_names.index(name)
            print(f"Found {name} at index {position}")
        else:
            print("Student not found.")

    elif choice == "4":
        student_names.sort()
        print("Sorted list:", student_names)
        print("Total students:", len(student_names))

    elif choice == "5":
        print("Final student list:", student_names)
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please enter 1-5.")
```
